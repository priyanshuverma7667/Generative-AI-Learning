```python
import nltk
from nltk.tokenize import word_tokenize
nltk.download('punkt_tab')

sentence = "Natural Language    processong with NLTK is fun and exciting!"
print(word_tokenize(sentence))
```

```python
from nltk.tokenize import sent_tokenize

print(sent_tokenize(sentence))
```

```python
from nltk.tokenize import WhitespaceTokenizer, SpaceTokenizer

print(WhitespaceTokenizer().tokenize(sentence))

print(SpaceTokenizer().tokenize(sentence))
```

# Stop Words --

Stop words are extremely common words (like "the", "is", "at", and "on") that carry very little meaning and are often filtered out during text processing.


```python
import nltk
from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize
nltk.download('punkt')
nltk.download('punkt_tab')   # Required by newer NLTK versions to fix LookupError
nltk.download('stopwords')



```

```python

text = 'Tokenization is essential for natural language processing.'

tokens = word_tokenize(text)

stop_words = set(stopwords.words('english'))

filtered_tokens = [word for word in tokens if word.lower() not in stop_words]

print(filtered_tokens)
```
