# Bag_Of_Word_(BOG)

corpus -->" I Love Machine Learning",
          " Machine Learning is fun"

D1 ---> ['I','Love','Machine','Learning']


*   List item
*   List item


D2 ---> ['Machine','Learning','is','fun']

Voc = I Love Machine Learning is fun

B_O_G_(D1) = 1 1 1 1 0 0
B_O_G_(D2) = 0 0 1 1 1 1


```python
from sklearn.feature_extraction.text import CountVectorizer
import pandas as pd

corpus = ["I Love Machine Learning","Machine Learning is fun"]

vectorizer = CountVectorizer()
x = vectorizer.fit_transform(corpus)

feature_names = vectorizer.get_feature_names_out()
feature_names

dense = x.todense()
dense

df = pd.DataFrame(dense,columns=feature_names)
df
```
