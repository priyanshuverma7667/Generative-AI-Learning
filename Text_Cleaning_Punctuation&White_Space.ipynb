## Text Cleaning Punctuation & White Space

### Removing Punctuation !

```python
filtered_token = ['Hello!','This ','is', 'a.','test,','let\'s','remove@','punctuation!']

import string

punctuation_removed = ["".join(char for char in word if char not in string.punctuation) for word in filtered_token]
print(punctuation_removed)
```

### Removing numbers ands special chars

```python
import re

def clear_text(text):
    cleaned_text =re.sub(r'[^A-Za-z\s]', '', text)
    return cleaned_text

text = "Hello! This is@ a test123, let's remove334 punctuation!"
clear_text(text)
```

### Removing whitespaces

```python
text_with_whitespaces = "Hey My name is Priyanshu     kumar   Verma & I am MCA   graduate."

#Remove Extra WhiteSpaces
text_without_whitespaces = " ".join(text_with_whitespaces.split())
print(text_without_whitespaces)
```
